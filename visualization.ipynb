```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Simple line plot
x = np.array([1, 2, 3, 4, 5, 6])
y = np.array([2, 4, 6, 8, 10, 12])

plt.figure(figsize=(6, 4))
plt.plot(x, y)
plt.title('Simple Line Plot', fontsize=16, fontweight='bold')
plt.xlabel('X Values', fontsize=12)
plt.ylabel('Y Values', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Multiple lines with styling
months = np. array([ ' Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'])
laptop = np.array([50, 55, 60, 58, 65, 70])
mobile = np.array([45, 48, 52, 55, 60, 58])
airbods = np.array([40, 42, 45, 50, 52, 55])
plt.figure(figsize=(8,6))
plt.plot(months, laptop, marker='o', linewidth=2, label='laptop', color='#667eea')
plt.plot(months, mobile, marker='s', linewidth=2, label='mobile', color='#ff6b6b')
plt.plot(months, airbods, marker='^', linewidth=2, label='airbods', color='#4ecdc4')

plt.title('Monthly Sales Comparison', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Sales (Units)', fontsize=12)
plt.legend(loc='upper left', fontsize=10)
plt.grid(True, alpha=0.3, linestyle='--')
plt.tight_layout()
plt.show()
```

```python
# Generate sample data - hours studied vs exam scores
np.random.seed(42)
hours_studied = np.random.uniform(0, 10, 50)
exam_scores = 40 + 5 * hours_studied + np.random.normal(0, 5, 50)

plt.figure(figsize=(10, 6))
plt.scatter(hours_studied, exam_scores, c='#667eea', alpha=0.6, s=100, edgecolors='black', linewidth=1)
plt.title('Relationship: Study Hours vs Exam Scores', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Hours Studied', fontsize=12)
plt.ylabel('Exam Score', fontsize=12)
plt.grid(True, alpha=0.3)

# Add trend line
z = np.polyfit(hours_studied, exam_scores, 1)
p = np.poly1d(z)
plt.plot(hours_studied, p(hours_studied), "r--", linewidth=2, label='Trend Line')
plt.legend()
plt.tight_layout()
plt.show() 
```

```python
# Generate sample data - exam scores
np.random.seed(42)
scores = np.random.normal(75, 10, 1000) # mean=75, std=10

plt.figure(figsize=(12, 5))

# Histogram
plt.subplot(1, 2, 1)
plt.hist(scores, bins=30, color='#667eea', alpha=0.7, edgecolor='black')
plt.title('Distribution of Exam Scores', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Score', fontsize=11)
plt.ylabel('Frequency', fontsize=11)
plt.axvline(scores.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean = {scores.mean() :.1f}')
plt.legend()
plt.grid(axis='y', alpha=0.3)

# Histogram with density (normalized)
plt.subplot(1, 2, 2)
plt.hist(scores, bins=30, color='#764ba2', alpha=0.7, edgecolor='black', density=True)
plt.title('Normalized Distribution', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Score', fontsize=11)
plt.ylabel('Density', fontsize=11)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
```

```python
# Market share data
labels = ['Python', 'Java', 'JavaScript', 'C#', 'Others']
sizes = [30, 25, 20, 15, 10]
colors = ['#667eea', '#764ba2', '#f093fb', '#4facfe', '#43e97b']
explode = (0.1, 0, 0, 0, 0) # Explode the first slice

plt.figure(figsize=(10, 8))
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%',
shadow=True, startangle=90, textprops={'fontsize': 12, 'fontweight': 'bold'})
plt.title('Programming Language Market Share', fontsize=16, fontweight='bold', pad=20)
plt.axis('equal') # Equal aspect ratio ensures circular pie
plt.tight_layout()
plt.show()
```

```python
# Donut chart
fig, ax = plt.subplots(figsize=(10, 8))
ax.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%',
startangle=90, pctdistance=0.85, textprops={'fontsize': 11, 'fontweight': 'bold'})

# Draw circle in center to create donut
centre_circle = plt.Circle((0, 0), 0.70, fc='white')
fig.gca().add_artist(centre_circle)

ax. set_title('Market Share - Donut Chart', fontsize=16, fontweight='bold', pad=20)
ax.axis('equal')
plt.tight_layout()
plt.show()
```

```python
# Generate comprehensive sales data

np.random.seed(42)
 
# Data

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']

revenue = [45000, 52000, 48000, 58000, 62000, 68000]

expenses = [30000, 33000, 32000, 36000, 38000, 40000]

profit = [r - e for r, e in zip(revenue, expenses)]
 
products = ['Electronics', 'Clothing', 'Food', 'Books', 'Others']

product_sales = [35, 25, 20, 12, 8]
 
customer_ages = np.random.normal(35, 12, 500)
 
# Create dashboard

fig = plt.figure(figsize=(16, 10))

fig.suptitle('📊 Sales Analytics Dashboard - Q1 & Q2 2024', 

             fontsize=20, fontweight='bold', y=0.98)
 
# 1. Revenue vs Expenses (Line Chart)

ax1 = plt.subplot(2, 3, 1)

ax1.plot(months, revenue, marker='o', linewidth=2.5, label='Revenue', color='#667eea')

ax1.plot(months, expenses, marker='s', linewidth=2.5, label='Expenses', color='#ff6b6b')

ax1.set_title('Revenue vs Expenses', fontsize=14, fontweight='bold', pad=10)

ax1.set_xlabel('Month')

ax1.set_ylabel('Amount ($)')

ax1.legend()

ax1.grid(True, alpha=0.3)
 
# 2. Monthly Profit (Bar Chart)

ax2 = plt.subplot(2, 3, 2)

bars = ax2.bar(months, profit, color='#4ecdc4', alpha=0.8, edgecolor='black')

ax2.set_title('Monthly Profit', fontsize=14, fontweight='bold', pad=10)

ax2.set_xlabel('Month')

ax2.set_ylabel('Profit ($)')

ax2.grid(axis='y', alpha=0.3)

for bar in bars:

    height = bar.get_height()

    ax2.text(bar.get_x() + bar.get_width()/2., height,

             f'${height:,.0f}', ha='center', va='bottom', fontsize=9)
 
# 3. Product Sales Distribution (Pie Chart)

ax3 = plt.subplot(2, 3, 3)

colors_pie = ['#667eea', '#764ba2', '#f093fb', '#4facfe', '#43e97b']

ax3.pie(product_sales, labels=products, colors=colors_pie, autopct='%1.1f%%',

        startangle=90, textprops={'fontsize': 9})

ax3.set_title('Sales by Category', fontsize=14, fontweight='bold', pad=10)
 
# 4. Customer Age Distribution (Histogram)

ax4 = plt.subplot(2, 3, 4)

ax4.hist(customer_ages, bins=30, color='#764ba2', alpha=0.7, edgecolor='black')

ax4.axvline(customer_ages.mean(), color='red', linestyle='--', linewidth=2, 

            label=f'Mean: {customer_ages.mean():.1f}')

ax4.set_title('Customer Age Distribution', fontsize=14, fontweight='bold', pad=10)

ax4.set_xlabel('Age')

ax4.set_ylabel('Frequency')

ax4.legend()

ax4.grid(axis='y', alpha=0.3)
 
# 5. Growth Rate (Line Chart with Fill)

ax5 = plt.subplot(2, 3, 5)

growth_rate = [(profit[i] - profit[i-1])/profit[i-1]*100 if i > 0 else 0 

               for i in range(len(profit))]

ax5.plot(months, growth_rate, marker='D', linewidth=2, color='#667eea')

ax5.fill_between(range(len(months)), growth_rate, alpha=0.3, color='#667eea')

ax5.axhline(y=0, color='black', linestyle='-', linewidth=0.8)

ax5.set_title('Month-over-Month Growth Rate', fontsize=14, fontweight='bold', pad=10)

ax5.set_xlabel('Month')

ax5.set_ylabel('Growth Rate (%)')

ax5.grid(True, alpha=0.3)
 
# 6. Key Metrics (Text Summary)

ax6 = plt.subplot(2, 3, 6)

ax6.axis('off')

metrics_text = f"""

KEY PERFORMANCE INDICATORS

{'='*35}
 
Total Revenue:      ${sum(revenue):>12,}

Total Expenses:     ${sum(expenses):>12,}

Total Profit:       ${sum(profit):>12,}
 
Avg Monthly Revenue: ${np.mean(revenue):>11,.0f}

Avg Monthly Profit:  ${np.mean(profit):>11,.0f}
 
Profit Margin:      {(sum(profit)/sum(revenue)*100):>12.1f}%
 
Best Month:         {months[np.argmax(profit)]:>16}

Best Profit:        ${max(profit):>12,}
 
Total Customers:    {len(customer_ages):>16,}

Avg Customer Age:   {customer_ages.mean():>16.1f}

"""

ax6.text(0.1, 0.5, metrics_text, fontsize=11, family='monospace',

         verticalalignment='center', bbox=dict(boxstyle='round', 

         facecolor='#f0f0f0', alpha=0.8))
 
plt.tight_layout()

plt.show()
 
```
