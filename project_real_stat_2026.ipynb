# IQR

```python
class OutlierAnalysis:

    def __init__(self, df, property_type):
        self.df = df
        self.property_type = property_type

        self.data = df[
            df["property_type"] == property_type
        ]["price_egp"].dropna()

    def describe(self):
        return self.data.describe()

    def iqr(self):
        Q1 = self.data.quantile(0.25)
        Q3 = self.data.quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = self.data[
            (self.data < lower) |
            (self.data > upper)
        ]

        print(f"Q1: {Q1:,.2f}")
        print(f"Q3: {Q3:,.2f}")
        print(f"IQR: {IQR:,.2f}")
        print(f"Lower Bound: {lower:,.2f}")
        print(f"Upper Bound: {upper:,.2f}")
        print(f"Outliers Count: {len(outliers):,}")

    def get_outliers(self):

        Q1 = self.data.quantile(0.25)
        Q3 = self.data.quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        result = self.df[
            (self.df["property_type"] == self.property_type) &
            (
                (self.df["price_egp"] < lower) |
                (self.df["price_egp"] > upper)
            )
        ][[
            "property_type",
            "price_egp",
            "area_value",
            "price_per_sqm",
            "city"
        ]].copy()

        return result.sort_values(
            "price_egp",
            ascending=False
        )
```

```python
def get_segment_outliers(df, property_type):

    # Segment
    segment = df[df["property_type"] == property_type].copy()

     
    Q1 = segment["price_egp"].quantile(0.25)
    Q3 = segment["price_egp"].quantile(0.75)

    IQR = Q3 - Q1

    outliers = segment[
        (segment["price_egp"] < Q1 - 1.5 * IQR) |
        (segment["price_egp"] > Q3 + 1.5 * IQR)
    ]

    print(f"Property Type: {property_type}")
    print(f"Q1: {Q1:,.2f}")
    print(f"Q3: {Q3:,.2f}")
    print(f"IQR: {IQR:,.2f}")
    print(f"Outliers Count: {len(outliers):,}")

    # إرجاع  Outliers
    return outliers[
        [
            "property_type",
            "price_egp",
            "area_value",
            "price_per_sqm",
            "city",
            "district"
        ]
    ].sort_values("price_egp", ascending=False)
```

```python
import pandas as pd
from scipy.stats import mannwhitneyu

# 1. تقسيم البيانات
furnished = dfr[dfr["furnished"] == "YES"]["price_egp"].dropna()
unfurnished = dfr[dfr["furnished"] == "NO"]["price_egp"].dropna()

# 2. حساب الـ Median
furnished_median = furnished.median()
unfurnished_median = unfurnished.median()

# 3. حساب إجمالي الـ Median
total_median = furnished_median + unfurnished_median

# 4. حساب نسبة كل مجموعة من إجمالي الـ Median
furnished_percentage = (furnished_median / total_median) * 100
unfurnished_percentage = (unfurnished_median / total_median) * 100

# 5. Mann–Whitney U Test
u_stat, p_value = mannwhitneyu(
    furnished,
    unfurnished,
    alternative="two-sided"
)

# 6. تحديد نتيجة الاختبار
alpha = 0.05

if p_value < alpha:
    significance = "Significant"
else:
    significance = "Not Significant"

# 7. إنشاء جدول المقارنة
comparison = pd.DataFrame({
    "Category": ["Furnished", "Unfurnished"],
    "Count": [len(furnished), len(unfurnished)],
    "Median Rent (EGP)": [
        furnished_median,
        unfurnished_median
    ],
    "Price Percentage": [
        furnished_percentage,
        unfurnished_percentage
    ]
})

# 8. تنسيق الأرقام
comparison["Median Rent (EGP)"] = comparison["Median Rent (EGP)"].round(2)
comparison["Price Percentage"] = comparison["Price Percentage"].round(2)

display(comparison)

# 9. عرض نتائج الاختبار في جدول منفصل
test_results = pd.DataFrame({
    "Test": ["Mann–Whitney U"],
    "U-statistic": [u_stat],
    "P-value": [p_value],
    "Alpha": [alpha],
    "Result": [significance]
})

display(test_results)



```

Insight: Furnished properties have a higher average rent than unfurnished properties, and the difference is statistically significant (p-value < 0.05).

```python
import pandas as pd
from scipy.stats import kruskal

# 1. Prepare the data
df_test = dfr[["price_egp", "city", "furnished"]].dropna()

# Keep cities with at least 5 properties
city_counts = df_test["city"].value_counts()
valid_cities = city_counts[city_counts >= 5].index

df_test = df_test[df_test["city"].isin(valid_cities)]

# 2. City Summary Table
location_summary = (
    df_test
    .groupby("city")
    .agg(
        Count=("price_egp", "count"),
        Median_Rent=("price_egp", "median"),
        Furnished_Count=("furnished", lambda x: (x == "YES").sum()),
        Unfurnished_Count=("furnished", lambda x: (x == "NO").sum())
    )
    .reset_index()
)

# 3. Calculate percentages
location_summary["Furnished_Percentage"] = (
    location_summary["Furnished_Count"]
    / location_summary["Count"]
) * 100

location_summary["Unfurnished_Percentage"] = (
    location_summary["Unfurnished_Count"]
    / location_summary["Count"]
) * 100

# 4. Round values
location_summary["Median_Rent"] = location_summary["Median_Rent"].round(2)

location_summary["Furnished_Percentage"] = (
    location_summary["Furnished_Percentage"].round(2)
)

location_summary["Unfurnished_Percentage"] = (
    location_summary["Unfurnished_Percentage"].round(2)
)

# 5. Sort by Furnished Percentage
location_summary = location_summary.sort_values(
    "Furnished_Percentage",
    ascending=False
)

print("City Summary (RENT)")
display(location_summary)

# 6. Kruskal-Wallis H Test
groups = [
    group["price_egp"].values
    for name, group in df_test.groupby("city")
]

h_stat, p_value = kruskal(*groups)

# 7. Test Results Table
test_results = pd.DataFrame({
    "Test": ["Kruskal-Wallis H Test"],
    "H-Statistic": [h_stat],
    "P-Value": [p_value],
    "Alpha": [0.05],
    "Decision": [
        "Reject H0" if p_value < 0.05
        else "Fail to Reject H0"
    ]
})

print("\nKruskal-Wallis Results (RENT)")
display(test_results)
```

Insight: The One-Way ANOVA test shows that the average rent differs significantly across cities (p-value < 0.05). This indicates that location is associated with differences in rental prices.
